```python
# 4. Download Model Weights from Hugging Face (~7.1 GB)
from pathlib import Path
from huggingface_hub import snapshot_download

model_dir = Path('/kaggle/working/breeze-tts-2')
if not (model_dir / 'config.json').exists():
    print("Downloading Breeze TTS 2 weights from Hugging Face (breezeblue-ai/breeze-tts-2)...")
    snapshot_download('breezeblue-ai/breeze-tts-2', local_dir=str(model_dir))
    print("Weights download complete!")
else:
    print(f"Weights already found at {model_dir}")
```

```python
# 5. Launch Dual-GPU Gradio Web UI with Public Share Link
import sys
repo_path = '/kaggle/working/breeze-tts-2-gpu-t4-optimized'
if repo_path not in sys.path:
    sys.path.insert(0, repo_path)

from app import create_ui, get_engine

print("Initializing Dual-GPU Breeze TTS Engine across cuda:0 and cuda:1...")
get_engine(model_dir=str(model_dir))

demo = create_ui()
demo.queue().launch(share=True, inline=True)
```
